1. What is the difference between a function and a method in Python?

A function is an independent block of reusable code defined with def. It is called by its name and is not
tied to any object. A method is a function that is defined inside a class and belongs to an object; it is called
using the dot notation (object.method()) and automatically receives the object as its first argument
(self).

```python
def greet(name): # function
 return "Hello " + name


class Person:
 def __init__(self, name):
  self.name = name
 def greet(self): # method
  return "Hello " + self.name



print(greet("Amit")) # function call
print(Person("Amit").greet()) # method call
```

5. What are iterators in Python and how do they differ from iterables?

An iterable is any object that can return an iterator, i.e. it implements __iter__(). Examples: list, tuple,
string, set, dictionary. An iterator is an object that produces the next value one at a time. It implements
__iter__() and __next__() and raises StopIteration when no values remain.

```python
data = [10, 20, 30]
it = iter(data)
print(next(it))
print(next(it))
print(next(it))
try:
 next(it)
except StopIteration:
 print("Iterator exhausted")

```

6. Explain the concept of generators in Python and how they are defined.

A generator is a special kind of iterator that produces values lazily, one at a time, only when requested.
Instead of return, it uses the yield keyword. When yield is executed the function pauses and keeps
its local state; the next call to next() resumes from where it stopped.

7. What are the advantages of using generators over regular functions?


• Memory efficient: values are produced one at a time, so the full sequence is never stored in memory.


• Lazy evaluation: a value is computed only when needed, which saves time if not all values are used.


• Can represent infinite sequences (e.g. endless Fibonacci numbers) which a list cannot.
• Cleaner code: no need to write a full iterator class with __iter__ and __next__.
• Preserve state automatically between calls (local variables are kept).
• Good for pipelines and large files: data can be processed in a stream, line by line.


8. What is a lambda function in Python and when is it typically used?

A lambda function is a small anonymous (unnamed) function written in a single line using the syntax
lambda arguments: expression. It can take any number of arguments but contains only one
expression, whose value is returned automatically.


```python
sq=lambda a: a*a
sq(4)
print((lambda x: x ** 2)(5))
words = ["pear", "fig", "banana"]
print(sorted(words, key=lambda w: len(w)))
```

9. Explain the purpose and usage of the map() function in Python

map(function, iterable, ...) applies the given function to every item of one or more iterables and
returns a map object (an iterator). Convert it with list() to see the results. It avoids writing explicit loops
and is commonly combined with lambda.

```python
nums=[1,2,3,4]
print(list(map(lambda a: a*a,nums)))
print(list(map(lambda a, b: a + b, [1, 2, 3], [10, 20, 30])))
print(list(map(str.upper,["a","b"])))
```

10. What is the difference between map(), reduce(), and filter() functions in Python?

Function	Purpose	Output
map()	Applies a function to every element.
An iterator containing transformed elements.

filter()	Selects elements that satisfy a condition.	An iterator containing selected elements.

reduce()	Combines all elements into a single value.	A single accumulated result.

```python
from functools import reduce
nums=(1,5,7,9,3)

print(list(map(map(lambda a: a**2,nums))))
print(list(filter(lambda b: b%2==0,nums)))
print(list(reduce(lambda x,y:x+y,nums)))
```
